```python
import sys
import logging
import numpy as np
import xarray as xr
import pandas as pd

import holoviews as hv
import colorcet as cc
from holoviews import opts

from tqdm_joblib import ParallelPbar
from joblib import delayed

from gnss_tid.pointdata import PointData
from gnss_tid.old_pointdata import PointData as old_PointData

hv.extension("bokeh")
opts.defaults(
    opts.Image(data_aspect=1, frame_height=300, frame_width=500, cmap=cc.cm.coolwarm, colorbar=True, show_grid=True, tools=["hover"]),
    hv.opts.Scatter(data_aspect=1, frame_height=300, frame_width=500, cmap=cc.cm.coolwarm, colorbar=True, show_grid=True, tools=["hover"]),
)
logging.basicConfig(stream=sys.stdout)
logging.root.setLevel(logging.INFO)
```

```python
time_limits = ["20150325_234000", "20150325_235200"]
latitude_limits = [24, 48]
longitude_limits = [-120, -75]
TI = 5
HEIGHT = 200
WINDOW = 4
STEP = 2
```

```python
print("LOADING NEW DATA")

new = {}
new["points"] = PointData(
  "/disk1/tid/2015/0325/*.nc",
  latitude_limits=latitude_limits,
  longitude_limits=longitude_limits,
  time_limits=time_limits,
  el_min=30,
)
new["time_slices"], new["times"] = new["points"].get_time_slices(WINDOW, STEP)
```

```python
hours_in_files = 30
minutes = 12.5

print(f"expected time ratio = {minutes/(hours_in_files*60)}")
valid = new["points"].debug_table.valid_sum.sum()
n_sv = new["points"].debug_table.unique_valid_svs.apply(lambda x: x.size)
valid_per_sv = valid / n_sv.sum()
print(f"{valid_per_sv = }")
print(f"valid per sv per min = {valid_per_sv / minutes}")

time_valid_ratio_expected = (new["points"].debug_table.valid_time_mean >= (minutes/(hours_in_files*60))).mean()
print(f"{time_valid_ratio_expected = }")

all_svs = np.unique(np.concat(new["points"].debug_table.unique_svs.values))
filtered_svs = np.unique(np.concat(new["points"].debug_table.unique_valid_svs.values))
missing_svs = all_svs[~np.isin(all_svs, filtered_svs)]
print("num svs within time limits ", filtered_svs.shape[0])
print("num svs in all files ", all_svs.shape[0])

print("any missing time/el/az in any file?", new["points"].debug_table.fin_mask_mean.min() < 1)


for v in ["dtec0", "dtec1", "dtec2", "dtec3", "dtecp"]:
    null_sum = new["points"].debug_table[f"{v}_null_sum"].sum()
    print(f"{v} null sum = {null_sum}")
    print(f"{v} null rate = {null_sum / valid}")

new["points"].debug_table.drop(columns=["unique_valid_svs", "unique_svs"])
```

```python
print("LOADING OLD DATA")

old = {}
old["points"] = old_PointData(
    ["/disk1/tid/data/2015_0325T0000-0326T0000_all0325.yaml_30el_30s_ra.h5"],
    latitude_limits=latitude_limits,
    longitude_limits=longitude_limits,
    time_limits=time_limits,
)
old["time_slices"], old["times"] = old["points"].get_time_slices(WINDOW, STEP)
```

```python
new["data"] = new["points"].get_data(new["time_slices"][TI], HEIGHT)
new["data"]
```

```python
old["data"] = old["points"].get_data(old["time_slices"][TI], HEIGHT)
old["data"]
```

```python
svs_old, counts_old = np.unique_counts(old["data"].prn)
svs_new, counts_new = np.unique_counts(new["data"].sv)

(
    hv.Bars(
        (svs_old, counts_old), kdims="prn"
    ).opts(width=500, height=400, show_grid=True, bar_width=1.5).relabel("old")
    + hv.Bars(
        (svs_new, counts_new), kdims="sv"
    ).opts(width=500, height=400, show_grid=True).relabel("new")
)
```

```python
from gnss_tid.image import ScipyRbfImageMaker

TECVAR = "dtec1"

image_maker = ScipyRbfImageMaker(
    hres=20,
    hp_freq=.02,
    neighbor_radius=100.0,
    neighbors=50,
    kernel="multiquadric",
    smoothing=.2,
)
b = {
    "west": -2224.88,
    "south": -1326.55,
    "east": 2231.40,
    "north": 1614.67,
}
image_maker.initialize(old["data"].x.values, None, b)


old["img"] = image_maker(old["data"].x.values, old["data"].y.values, old["data"].tec.values)
new["img"] = image_maker(new["data"].x.values, new["data"].y.values, new["data"][TECVAR].values)

old_pts = np.isfinite(old["data"].tec.values).sum()
new_pts = np.isfinite(new["data"][TECVAR].values).sum()

cmap = cc.cm.diverging_bwr_20_95_c54
(
    hv.Scatter(
        old["data"].to_dataframe(), kdims="x", vdims=["y", "tec"]
    ).opts(size=5, clim=(-.4, .4), cmap=cmap, symmetric=True, color="tec").relabel(f"old: N={old_pts}")
    + hv.Scatter(
        new["data"].to_dataframe(), kdims="x", vdims=["y", TECVAR]
    ).opts(size=5, clim=(-.4, .4), cmap=cmap, symmetric=True, color=TECVAR).relabel(f"new {TECVAR}: N={new_pts}")
    + hv.Image(
        old["img"].image.where(old["img"].density > 0), kdims=["x", "y"], vdims="tec"
    ).opts(cmap=cmap, symmetric=True)
    + hv.Image(
        new["img"].image.where(new["img"].density > 0), kdims=["x", "y"], vdims=TECVAR
    ).opts(cmap=cmap, symmetric=True)
).cols(2)
```
