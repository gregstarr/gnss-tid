```python
import logging
import numpy as np
import holoviews as hv
import xarray
import numpy as np
import pandas as pd
import colorcet as cc
from tqdm_joblib import ParallelPbar
from joblib import delayed
hv.extension("bokeh")
from holoviews import opts

import gnss_tid.pointdata
import gnss_tid.image

opts.defaults(
    opts.Image(data_aspect=1)
)
opts.defaults(
    opts.Image(data_aspect=1, frame_height=300, frame_width=500, cmap=cc.cm.coolwarm, colorbar=True, show_grid=True, tools=["hover"]),
    hv.opts.Scatter(data_aspect=1, frame_height=300, frame_width=500, cmap=cc.cm.coolwarm, colorbar=True, show_grid=True, tools=["hover"]),
)
logging.basicConfig(level=logging.INFO)
```

```python
from gnss_tid.utils import normalize_paths
files = normalize_paths("/disk1/tid/sharon/poly/2015/0325/*.nc")
NAMES = ['mobu', 'mobw', 'mocd', 'moce', 'moch', 'mocl', 'mocn', 'moco',
       'mocr', 'modp', 'modr', 'modx', 'moeb', 'moed', 'moel', 'mofe',
       'mogf', 'mohl', 'mohs', 'moht', 'mohv', 'moke', 'mokv', 'momc',
       'moml', 'momv', 'mone', 'moof', 'mopn', 'morm', 'mosd', 'mosf',
       'mosg', 'mosh', 'mosl', 'mosr', 'most', 'movb', 'mowb', 'mowr',
       'mows', 'moal', 'mobf', 'mobg', 'mobo', 'mobr', 'mobt', 'sg01',
       'sg04', 'sg45', 'sg47', 'sa30', 'txcd', 'txge', 'mopl']
files = list(filter(lambda x: x.stem.split("_")[0] in NAMES, files))
f = xarray.open_dataset(files[1])
```

```python
plot = None
ii = 13
N = 12
for i, file in enumerate(files[ii:ii+N]):
    print(file)
    f = xarray.open_dataset(file)
    p = hv.Points(
        (f.time.values, f.dtec0.values, f.sv.values), 
        kdims=["time", "dtec0"], vdims=["sv"]
    )
    if i > N-4:
        p = p.opts(color="sv", cmap=cc.cm.glasbey, width=400, height=200, title=f.filename, show_legend=False, show_grid=True)
    else:
        p = p.opts(color="sv", cmap=cc.cm.glasbey, width=400, height=200, title=f.filename, show_legend=False, xticks=0, xlabel="", show_grid=True)
    if plot is None:
        plot = p
    else:
        plot += p

plot.cols(3)
```

```python
time_limits = ["20150325_234000", "20150326_001500"]
latitude_limits = [24, 48]
longitude_limits = [-120, -75]
TI = 5
HEIGHT = 200
WINDOW = 4
STEP = 2

points = gnss_tid.pointdata.PointData(
  "/disk1/tid/2015/0325/*.nc",
  latitude_limits=latitude_limits,
  longitude_limits=longitude_limits,
  time_limits=time_limits,
  el_min=30,
)
time_slices, times = points.get_time_slices(4, 2)
```

```python
with ParallelPbar("getting data")(n_jobs=24) as pool:
    data = pool(delayed(points.get_data)(t, 200) for t in time_slices)
# data = xarray.concat(data, "height")
```

```python
hv.HoloMap(
    {
        t: hv.Scatter(d.to_dataframe(), kdims="x", vdims=["y", "dtec1"]).opts(size=3, clim=(-.4, .4), cmap="coolwarm", symmetric=True, color="dtec1", frame_height=500, frame_width=600)
        for t, d in zip(times, data)
        }
)

```

```python
image_maker1 = gnss_tid.image.ScipyRbfImageMaker(
    hres=20,
    hp_freq=.02,
    neighbor_radius=100.0,
    neighbors=50,
    kernel="multiquadric",
    smoothing=.4,
)

image_maker2 = gnss_tid.image.MetpyImageMaker(
    hres=20,
    hp_freq=.02,
    neighbor_radius=100.0,
    interp_type="rbf",
    rbf_func="multiquadric",
    rbf_smooth=.2,
)
```

```python
vname = "dtec1"
d = data[10]
print(d.height)
img1 = image_maker1(d.x.values, d.y.values, d[vname].values)
img2 = image_maker2(d.x.values, d.y.values, d[vname].values)
```

```python
vname = "dtec1"

d = points.get_data(time_slices[0], 210)

@delayed
def img(n, s):
    logging.basicConfig(level=logging.INFO)
    image_maker = gnss_tid.image.ScipyRbfImageMaker(
        hres=20,
        hp_freq=.02,
        neighbor_radius=100.0,
        neighbors=n,
        kernel="multiquadric",
        smoothing=s,
    )
    img = image_maker(d.x.values, d.y.values, d[vname].values)
    hv.extension("bokeh")
    return (-1 if n is None else n, s), hv.Image(
        img.image.where(img.density >= 5),
        kdims=["x", "y"],
    ).opts(cmap=cc.cm.coolwarm, colorbar=True).relabel(np.datetime_as_string(times, unit="m")[0])


with ParallelPbar("making images")(n_jobs=20) as pool:
    results = pool(img(n, s) for n in [10, 20, 50, 100] for s in [.1, .2, .3, .4, .5, .6])
```

```python
hv.HoloMap({k: v for k, v in results}, kdims=["neighbors", "smoothing"]).opts(frame_height=400, colorbar=True, clim=(-.3, .3), cmap=cc.cm.coolwarm)
```
