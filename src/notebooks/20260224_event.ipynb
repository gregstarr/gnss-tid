```python
import sys
import logging
import numpy as np
import xarray as xr
import pandas as pd

import holoviews as hv
import colorcet as cc
from holoviews import opts

from tqdm_joblib import ParallelPbar
from joblib import delayed

from gnss_tid.pointdata import PointData
from gnss_tid.pointdata import PointData as old_PointData

hv.extension("bokeh")
opts.defaults(
    opts.Image(data_aspect=1, frame_height=300, frame_width=500, cmap=cc.cm.coolwarm, colorbar=True, show_grid=True, tools=["hover"]),
    hv.opts.Scatter(data_aspect=1, frame_height=300, frame_width=500, cmap=cc.cm.coolwarm, colorbar=True, show_grid=True, tools=["hover"]),
    hv.opts.Curve(height=250, frame_width=350, show_grid=True, tools=["hover"]),
)
logging.basicConfig(stream=sys.stdout)
logging.root.setLevel(logging.INFO)
```

```python
data = xr.open_dataset("outputs/20241206/autofocus.h5")
params = xr.open_zarr("outputs/20241206/params.zarr")
params["anisotropy"] = abs(params["S2"]) / params["S0"]
params["power_snr"] = params["power_signal"] / params["power_noise"]
```

```python
params
```

```python
x = 150
y = 0
D = 20
L = 1200

r = np.arange(0, L, D)

def plot(theta, ti):
    img = data.image.isel(time=ti).where(data.density.isel(time=ti) > 0)
    parm = params.isel(time=ti)
    xline = xr.DataArray(r * np.cos(theta * np.pi / 180), dims="r", coords={"r": r})
    yline = xr.DataArray(r * np.sin(theta * np.pi / 180), dims="r", coords={"r": r})
    linedata = img.sel(x=xline, y=yline, method="nearest")
    lineparm = parm.sel(px=xline, py=yline, method="nearest")
    return (
        hv.Image(img, kdims=["x", "y"]).opts(title=f"{img.time.dt.strftime('%D %H:%M:%S').item()}") * hv.Points((xline.values, yline.values))
        + hv.Curve(linedata, kdims="r", vdims="image")
        + hv.Curve(lineparm, kdims="r", vdims="freq_snr")
        + hv.Curve(lineparm, kdims="r", vdims="anisotropy")
        + hv.Curve(lineparm, kdims="r", vdims="sigpower_signalnal")
        + hv.Curve(lineparm, kdims="r", vdims="period")
    )

thetas = np.arange(220, 260, 10)
times = np.arange(90, 109, 2)
hv.HoloMap(
    {
        (theta, ti): plot(theta, ti) for theta in thetas for ti in times
    },
    kdims=["theta", "time"]
).collate().cols(1)
```

```python
x = 150
y = 0
D = 20
L = 1200
theta = 230

r = np.arange(0, L, D)

img = data.image.where(data.density > 0)
xline = xr.DataArray(r * np.cos(theta * np.pi / 180), dims="r", coords={"r": r})
yline = xr.DataArray(r * np.sin(theta * np.pi / 180), dims="r", coords={"r": r})
linedata = img.interp(x=xline, y=yline)
lineparm = params.interp(px=xline, py=yline)

(
    hv.Image(linedata).opts(data_aspect=None).relabel(f"theta={theta}")
    + hv.Image(lineparm, vdims="anisotropy").opts(data_aspect=None, cmap=cc.cm.gouldian).relabel("anisotropy")
    + hv.Image(lineparm, vdims="power_snr").opts(clim=(0, 20), data_aspect=None, cmap=cc.cm.gouldian).relabel("power_snr")
    + hv.Image(lineparm, vdims="power_signal").opts(data_aspect=None, cmap=cc.cm.gouldian).relabel("power_signal")
    + hv.Image(lineparm, vdims="power_threshold").opts(data_aspect=None, cmap=cc.cm.gouldian).relabel("power_threshold")
).cols(1)
```
