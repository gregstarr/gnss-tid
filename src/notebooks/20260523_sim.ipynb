```python
import numpy as np
from gnss_tid.simulation_3d import SimulationScenario, ShellGrid
from gnss_tid.pointdata import project_to_ipp, lookup_rx_location
from datetime import datetime
import holoviews as hv
hv.extension("bokeh")
```

```python
grid = ShellGrid(
    res_lat=.1,
    res_lon=.1,
    lat_range=[24, 48],
    lon_range=[-120, -75],
)
grid.add_rectangle(300, (30, 32), (-100, -98), 1)
sim = SimulationScenario(grid, nav_file="../../tests/data/BRDC00IGS_R_20253500000_01D_MN.rnx")
sim.generate_random_receivers(2000, [30, 42], [-110, -85])
obs = sim.generate_observations_time_series(
    datetime(2025, 12, 16, 0, 0, 0),
    datetime(2025, 12, 16, 0, 10, 0),
    60,
)
obs = obs[obs.el > 30]
rx = sim.generate_receiver_df()
rxp = lookup_rx_location(obs, rx)
```

```python
proj_obs1 = project_to_ipp(obs, rxp[["lat", "lon", "alt"]].values, 200).sort_values("time")
proj_obs2 = project_to_ipp(obs, rxp[["lat", "lon", "alt"]].values, 300).sort_values("time")
(
    hv.Scatter(proj_obs1[np.isin(proj_obs1.time, proj_obs1.time.unique()[:5])], kdims=["lon"], vdims=["lat", "stec"]).opts(color="stec", colorbar=True, height=400, width=600) +
    hv.Scatter(proj_obs2[np.isin(proj_obs2.time, proj_obs2.time.unique()[:5])], kdims=["lon"], vdims=["lat", "stec"]).opts(color="stec", colorbar=True, height=400, width=600)
).cols(1)
```
