```python
df = r.to_dataframe()
df["patch_freq_snr"] = np.log10(df["patch_freq_snr"])
nzdf = df[df["r"] > 0]

# nzdf["period_bin"] = pd.cut(nzdf["period"], bins=6).apply(lambda x: (x.left + x.right) / 2)
# nzdf["phase_speed_bin"] = pd.cut(nzdf["phase_speed"], bins=6).apply(lambda x: (x.left + x.right) / 2)
# nzdf["wavelength_bin"] = pd.cut(nzdf["wavelength"], bins=6).apply(lambda x: (x.left + x.right) / 2)
nzdf["period_bin"] = pd.cut(nzdf["period"], bins=5)
nzdf["phase_speed_bin"] = pd.cut(nzdf["phase_speed"], bins=5)
nzdf["wavelength_bin"] = pd.cut(nzdf["wavelength"], bins=5)
```

```python
from sklearn.preprocessing import QuantileTransformer
from sklearn.decomposition import PCA
from sklearn.pipeline import Pipeline

X = nzdf[["wavelength", "anisotropy_mag"]]
Xp = Pipeline([
    ("A", PCA(whiten=True)),
    ("B", QuantileTransformer()),
]).fit_transform(X)

hv.HexTiles(
    (Xp[:, 0], Xp[:, 1], np.log10(abs(nzdf["rel_wavelength_error"]))),
    kdims=["wavelength", "anisotropy_mag"], vdims="rel_wavelength_error"
).opts(
    colorbar=True,
    cmap="gouldian",
    aggregator=np.median
)
```

```python
from sklearn.preprocessing import PolynomialFeatures, StandardScaler, QuantileTransformer
from sklearn.decomposition import PCA
from sklearn.pipeline import Pipeline
from sklearn.linear_model import Lasso, TweedieRegressor, RANSACRegressor

models = {}
# m = nzdf.loc[:, "lam"] < 500
for xnames in [("wavelength", "anisotropy_mag"), ("wavelength", "patch_freq_snr"), ("period", "patch_freq_snr"), ("period", "anisotropy_mag"), ("period", "power_threshold"), ("wavelength", "power_threshold")]:
    for yname in ["rel_wavelength_error", "rel_period_error"]:
        X = nzdf[list(xnames)].values
        y = np.log10(abs(nzdf[yname])).values
        # y = abs(nzdf[yname]).values
        models[(xnames, yname)] = Pipeline([
            ("scalar", StandardScaler()),
            # ("quantile", QuantileTransformer()),
            ("white", PCA(whiten=True)),
            ("poly", PolynomialFeatures(3)),
            ("reg", Lasso(alpha=.005)),
            # ("reg", RANSACRegressor(Lasso(alpha=.005), min_samples=.1))
            # ("reg", GammaRegressor(alpha=.1))
        ]).fit(X, y)
        print((xnames, yname))
        # for n, v in zip(models[(xnames, yname)].steps[1][1].get_feature_names_out(), models[(xnames, yname)].steps[-1][1].coef_):
        #     print(n, ":", v)
        # print()

Xall = nzdf[["wavelength", "period", "anisotropy_mag", "patch_freq_snr", "power_threshold"]]
yall = np.log10(abs(nzdf[["rel_wavelength_error", "rel_period_error"]])).values

M = Pipeline([
    ("scalar", StandardScaler()),
    # ("quantile", QuantileTransformer()),
    ("white", PCA(whiten=True)),
    ("poly", PolynomialFeatures(3)),
    ("reg", Lasso(alpha=.01)),
    # ("reg", RANSACRegressor(Lasso(alpha=.005), min_samples=.1))
    # ("reg", GammaRegressor(alpha=.1))
]).fit(Xall, yall)
out = M.predict(Xall)

hv.HexTiles((out[:, 1], yall[:, 1])) * hv.Path(([-3, 0], [-3, 0]))
```

```python
SNRVARS = (
    ["anisotropy_mag", "patch_freq_snr", "power_threshold"],
    ["anisotropy_bin", "freq_snr_bin", "power_bin"]
)
Z = "wavelength"
Y = "rel_period_error"
P = None
for X, B in zip(*SNRVARS):
    agg = nzdf.groupby(
        pd.qcut(nzdf[X], q=10).apply(lambda x: (x.left + x.right) / 2),
        observed=False,
    )[Y].agg(
        med=lambda s: np.median(abs(s)),
        p0=lambda s: np.percentile(abs(s), 2.5),
        p1=lambda s: np.percentile(abs(s), 97.5)
    ).reset_index()
    p = hv.Scatter(
        (nzdf.loc[::5, X], abs(nzdf.loc[::5, Y])), kdims=X, vdims=Y
    ).opts(
        frame_width=350,
        frame_height=350,
        show_grid=True,
        logy=True,
        size=1,
        alpha=.5,
    )
    p *= hv.Area((agg[X], agg["p0"], agg["p1"]), kdims=X, vdims=[Y, "top"]).opts(
        fill_alpha=.3,
        line_dash="dashed",
        line_width=1,
    )
    p *= hv.Curve((agg[X], agg["med"]), kdims=X, vdims=Y).opts(color="k")
    
    if ((Z, X), Y) in models:
        Q = {}
        groups = nzdf.groupby(f"{Z}_bin", observed=False)
        for z, d in groups:
            w = (z.left + z.right) / 2
            v = np.linspace(nzdf[X].min(), nzdf[X].max(), 10)
            x = np.column_stack((np.full_like(v, w), v))
            zz = 10**models[((Z, X), Y)].predict(x)
            Q[w] = hv.Curve((v, zz), kdims=X, vdims=Y).opts(
                color=hv.Cycle(colorcet_cmap_to_palette("gouldian", len(groups))),
                bgcolor="#CCCCCC",
                line_dash="dashed",
            )
        p *= hv.NdOverlay(Q, kdims=Z).opts(
            frame_height=350,
            frame_width=350,
            show_grid=True,
            logy=True,
            legend_position="right",
        )

    
    if P is None:
        P = p
    else:
        P += p
P
```
